```python
import heapq

def greedy_best_first_search(graph, heuristic, start, goal):

    # Priority queue: (heuristic value, node, path)
    queue = [(heuristic[start], start, [start])]

    visited = set()

    while queue:

        # Get node with smallest heuristic value
        h, node, path = heapq.heappop(queue)

        # Goal test
        if node == goal:
            return path

        # Skip if already visited
        if node in visited:
            continue

        visited.add(node)

        # Explore neighbours
        for neighbour in graph[node]:

            if neighbour not in visited:

                new_path = path + [neighbour]

                # Add neighbour using heuristic value
                heapq.heappush(
                    queue,
                    (heuristic[neighbour], neighbour, new_path)
                )

    return None


# Graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

# Heuristic values
heuristic = {
    'A': 40,
    'B': 25,
    'C': 20,
    'D': 35,
    'E': 30,
    'F': 10,
    'G': 0
}

# Start and goal
start = 'A'
goal = 'G'

# Run Greedy Best-First Search
path = greedy_best_first_search(graph, heuristic, start, goal)

print("Path:", path)
```

```python
def dfs(graph, start):

    stack = [start]
    visited_nodes = []

    while stack:

        node = stack.pop()

        if node not in visited_nodes:
            print(node, end=" ")
            visited_nodes.append(node)

            for neighbour in reversed(graph[node]):
                if neighbour not in visited_nodes:
                    stack.append(neighbour)


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

visited_nodes = dfs(graph, 'A')
```

```python
from collections import deque

def bfs(graph, start):

    queue = deque([start])
    visited = set()

    while queue:

        node = queue.popleft()

        if node not in visited:
            print(node, end=" ")
            visited.add(node)

            for neighbour in graph[node]:
                if neighbour not in visited:
                    queue.append(neighbour)


# Graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

bfs(graph, 'A')
```
