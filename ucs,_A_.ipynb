```python
import heapq

def uniform_cost_search(graph, start, goal):
    # Priority queue contains (cost, node, path)
    queue = [(0, start, [start])]
#empty set is created to remember already proccsd nodes
    visited = set()

    while queue:
        # getting the node with the minimum cost
        # heapq.heappop() : Take out the cheapest route currently available.
        cost, node, path = heapq.heappop(queue)

        # If goal is reached
        if node == goal:
            return path, cost

        # Skip already visited nodes
        if node in visited:
            continue

        visited.add(node)

        # Explore neighbours
        for neighbour, edge_cost in graph[node]:
            if neighbour not in visited:

                new_cost = cost + edge_cost
                new_path = path + [neighbour]

                heapq.heappush(
                    queue,
                    (new_cost, neighbour, new_path)
                )

    return None, float("inf")


# Graph representation
graph = {
    'A': [('B', 2), ('C', 5)],
    'B': [('D', 6), ('G', 10)],
    'C': [('D', 2)],
    'D': [('G', 3)],
    'G': []
}

path, cost = uniform_cost_search(graph, 'A', 'G')

print("Path:", path)
print("Total Cost:", cost)
```

```python
import heapq

def a_star(graph, heuristic, start, goal):

    # Queue contains: (f_cost, g_cost, node, path)
    queue = [(heuristic[start], 0, start, [start])]

    visited = set()

    while queue:

        # Remove node with smallest f(n)
        f_cost, g_cost, node, path = heapq.heappop(queue)

        # Check if goal is reached
        if node == goal:
            return path, g_cost

        # Skip if already visited
        if node in visited:
            continue

        visited.add(node)

        # Explore neighbours
        for neighbour, edge_cost in graph[node]:

            if neighbour not in visited:

                # Actual cost from start
                new_g = g_cost + edge_cost

                # f(n) = g(n) + h(n)
                new_f = new_g + heuristic[neighbour]

                # Update path
                new_path = path + [neighbour]

                # Add to priority queue
                heapq.heappush(
                    queue,
                    (new_f, new_g, neighbour, new_path)
                )

    return None, float("inf")


# Weighted graph
graph = {
    'A': [('B', 2), ('C', 3)],
    'B': [('D', 4), ('G', 7)],
    'C': [('D', 2)],
    'D': [('G', 2)],
    'G': []
}


# Heuristic values
heuristic = {
    'A': 6,
    'B': 5,
    'C': 3,
    'D': 1,
    'G': 0
}



path, cost = a_star(graph, heuristic, 'A', 'G')

print("Path:", path)
print("Total Cost:", cost)
```
